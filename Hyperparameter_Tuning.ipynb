```python
# importing the dependencies
import numpy as np
import pandas as pd
import sklearn.datasets
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV


```

working on breast cancer dataset


```python
# loading dataset
breast_cancer_dataset = sklearn.datasets.load_breast_cancer()
```

```python
print(breast_cancer_dataset)
```

```python
# loading data to dataframe
data_frame = pd.DataFrame(breast_cancer_dataset.data, columns = breast_cancer_dataset.feature_names)
```

```python
data_frame.head()
```

```python
data_frame.shape
```

```python
data_frame.isnull().sum()
```

```python
data_frame['label'] = breast_cancer_dataset.target
```

separating features and target

```python
x = data_frame.drop(columns = 'label' , axis = 1)
y = data_frame['label']
```

```python
print(x)
```

```python
print(y)
```

```python
x = np.asarray(x)
y = np.asarray(y)
```

# gridsearchCV

gridsearchCV is used to determine the best parameter for our model

```python
# load the svc model
model = SVC(

)
```

```python
# hyperparameters

parameters = {
    'kernel': ['linear','poly','rbf','sigmoid'],
    'C':[1,5,10,20]
}
```

```python
# gridsearch
classifier = GridSearchCV(model,parameters , cv =5)
```

```python
# fitting data in model
classifier.fit(x,y)
```

```python
classifier.cv_results_
```

```python
# best parameter
best_parameter = classifier.best_params_
print(best_parameter)
```

```python
# determine highest accuracy
highest_accuracy = classifier.best_score_
print(highest_accuracy)
```

```python
# laoding the results to pandas dataframe
result = pd.DataFrame(classifier.cv_results_)
```

```python
result.head(

)
```

```python
grid_search_result = result[['param_C','param_kernel','mean_test_score']]
```

```python
grid_search_result
```

hifghest accuracy is 95.2

best parameter = {'C' :10 , 'kernel':'linear'}
