# Array Manupulation

```python
import numpy as np

array = np.random.randint(0,10,(2,3))
print(array)
print(array.shape)
```

```python
#transpose
trans = np.transpose(array)
print(trans)
print(trans.shape)
```

```python
#another method of transpose
trans2 = array.T
print(trans2)
print(trans2.shape)
```

```python
# reshaping array
a = np.random.randint(0,10,(2,3))
print(a)
print(a.shape)
```

```python
b = a.reshape(3,2)
print(b)
print(b.shape)
```
