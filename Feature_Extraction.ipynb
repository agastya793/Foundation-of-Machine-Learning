# feature extraction


the mapping from textual data to real valued vectors (numerical data) is called feature extraction.


# bag of words (BOW)

list of unique words in the textual corpus.

# term frequency- document frequency (TF-IDF)

to count the number of times each word appears in a document.

# tf-idf vectorizer

  1.   term  frequency = number of times term t appears in a document/ number of terms in a document


2.  inverse document frequency = log(N/n), where N is the number of document and n is the number of document a term t has appeared in.

the IDF value of a rare word is high, whereas the IDF of a frequent word is low.

TF-IDF value of a term = TF*IDF


THIS IS HOW WE CONVERT TEXT DATA TO NUMERICAL DATA.
