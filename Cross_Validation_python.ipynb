# importing the dependencies


```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score
```

# importing model

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
```

# data collection and processing

```python
# loading the csv file to pandas dataframe
heart_data = pd.read_csv('/content/heart_disease_data.csv')
```

```python
heart_data.head()
```

```python
heart_data.isnull().sum()
```

```python
# check distribution of target variables
heart_data['target'].value_counts()
```

1-----> defective heart

0-----> healthy

split features and target

```python
x = heart_data.drop(columns = 'target' , axis =1)
y = heart_data['target']
```

```python
print(x)
```

```python
print(y)
```

# train test split

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2 ,stratify =y,random_state =3)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

# comparing the performance of the models

```python
# list of models
models = [LogisticRegression(max_iter = 1000),SVC(kernel = 'linear'), KNeighborsClassifier(),RandomForestClassifier()]
```

```python
def compare_models_train_test():
 for model in models :
  model.fit(x_train,y_train)
  test_data_prediction = model.predict(x_test)
  accuracy = accuracy_score(y_test , test_data_prediction)
  print('accuracy score of the' , model ,'=' , accuracy)
```

```python
compare_models_train_test()
```

# cross validation

```python
cv_score_lr = cross_val_score(LogisticRegression(max_iter = 1000), X=x ,y=y , cv = 5)

print(cv_score_lr)
mean_accuracy_lr = sum(cv_score_lr)/len(cv_score_lr)

mean_accuracy_lr = round(mean_accuracy_lr*100)
mean_accuracy_lr = round(mean_accuracy_lr,2)
print(mean_accuracy_lr)
```

```python
cv_score_svc = cross_val_score(SVC(kernel = 'linear'), X=x ,y=y , cv = 5)
print('cross validation score of SVC =',cv_score_svc)
mean_accuracy_svc = sum(cv_score_svc)/len(cv_score_svc)

mean_accuracy_svc = round(mean_accuracy_svc*100)
mean_accuracy_svc = round(mean_accuracy_lr,2)
print(mean_accuracy_lr)
```

```python
cv_score_knn = cross_val_score(KNeighborsClassifier(), X=x ,y=y , cv = 5)
print('cross validation score of KNN =',cv_score_knn)
```

```python
cv_score_rf = cross_val_score(RandomForestClassifier(), X=x ,y=y , cv = 5)
print('cross validation score of Random Forest =',cv_score_rf)
```

# creating a function to compare the models

```python
# list of models
models = [LogisticRegression(max_iter = 1000),SVC(kernel = 'linear'), KNeighborsClassifier(),RandomForestClassifier()]
```

```python
def compare_models_cross_validation():
  for model in models :
    cv_score = cross_val_score(model,x,y,cv = 5)
    mean_accuracy = sum(cv_score)/len(cv_score)
    mean_accuracy_svc = mean_accuracy*1000
    mean_accuracy_svc = round(mean_accuracy,2)

    print(' cross validation for the ' ,model ,'=' ,cv_score)
    print('accuracy % of the ' , model , mean_accuracy )
    print('----------------')
```

```python
compare_models_cross_validation()
```
