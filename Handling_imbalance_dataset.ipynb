# imbalance dataset:

a dataset with an unequal class distribution.

```python
# importing the dependencies
import numpy as np
import pandas as pd
```

```python
# load the dataset to pandas dataframe
credit_card_data = pd.read_csv('/content/credit_data.csv')
```

```python
# first five row of dataframe
credit_card_data.head()
```

```python
credit_card_data.tail()
```

# determine the distribution of two classes

```python
credit_card_data['Class'].value_counts()
```

this is highly imbalance dataset.


0------> legit transaction

1------> fraudulent transaction

```python
# separating the legit and fraudulent transaction
legit = credit_card_data[credit_card_data.Class == 0]
fraud = credit_card_data[credit_card_data.Class == 1]
```

```python
print(legit.shape)
print(fraud.shape)
```

# under-sampling

buid a sample dataset containing similar distribution of legit and fraud transaction.


number of fraud transaction---> 398

```python
legit_sample = legit.sample(n=398)
```

```python
print(legit_sample.shape)
```

# concatenate two dataframes

```python
new_dataset = pd.concat([legit_sample, fraud] , axis = 0)
```
