# matrix addition

two matrices can be added only if they have same shape

```python
# creating two matrices

A = np.array([[2,3],[4,5]])

B = np.array([[30,70],[60,100]])

print(A)
print(B)
```

```python
A.shape
```

```python
B.shape
```

adding two matrix

```python
sum =  A +B
print(sum)
```

# matrix subtraction

```python
A = np.array([[2,3],[4,5]])

B = np.array([[30,70],[60,100]])

print(A)
print(B)
```

```python
difference = A -B
print(difference)
```

```python
#create two matices with random values

matrix_1 = np.random.randint(100,size =(3,3))
matrix_2 = np.random.randint(100,size =(3,3))
print(matrix_1)
print(matrix_2)
```

```python
diff = np.subtract(matrix_1 , matrix_2)
```

```python
print(diff)
```
